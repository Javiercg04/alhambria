```python
import os, zipfile
from google.colab import files

carpeta_checkpoints = "/content/checkpoints_a_exportar"
os.makedirs(carpeta_checkpoints, exist_ok=True)

print("Sube el zip que contiene el/los checkpoint(s):")
subido = files.upload()
nombre_zip = list(subido.keys())[0]
with zipfile.ZipFile(nombre_zip, 'r') as zip_ref:
    zip_ref.extractall(carpeta_checkpoints)

checkpoints_disponibles = sorted(
    [f for f in os.listdir(carpeta_checkpoints) if f.startswith("checkpoint-")],
    key=lambda x: int(x.split("-")[1])
)
print("Checkpoints disponibles:", checkpoints_disponibles)
```

```python
!pip install --upgrade --force-reinstall --no-cache-dir unsloth unsloth_zoo
```

```python
import unsloth
import torch

from unsloth import FastModel
import shutil, gc

STEP_A_EXPORTAR = 2700

checkpoint_path = f"{carpeta_checkpoints}/checkpoint-{STEP_A_EXPORTAR}"

modelo, tokenizer = FastModel.from_pretrained(
    model_name = checkpoint_path,
    max_seq_length = 2048,
    load_in_4bit = True,
    load_in_8bit = False,
    full_finetuning = False,
)

gguf_dir = f"/content/gguf_step_{STEP_A_EXPORTAR}"
modelo.save_pretrained_gguf(gguf_dir, tokenizer, quantization_method="q4_k_m")

print(f"Exportación del checkpoint-{STEP_A_EXPORTAR} completada")
```

```python
print(gguf_dir)
```

```python
carpeta_real = f"{gguf_dir}_gguf"

ggufs = [f for f in os.listdir(carpeta_real) if f.endswith(".gguf")]
print("Archivos GGUF encontrados:", ggufs)

for f in ggufs:
    if "mmproj" not in f:
        os.rename(
            os.path.join(carpeta_real, f),
            os.path.join(carpeta_real, f"ms-{STEP_A_EXPORTAR}.gguf")
        )
        print(f"Renombrado a: ms-{STEP_A_EXPORTAR}.gguf")

zip_name = f"ms-{STEP_A_EXPORTAR}"
shutil.make_archive(zip_name, 'zip', carpeta_real)
files.download(f"{zip_name}.zip")
print(f"Descarga iniciada: {zip_name}.zip")
```

```python
shutil.rmtree(carpeta_real, ignore_errors=True)
shutil.rmtree(gguf_dir, ignore_errors=True)
del modelo, tokenizer
gc.collect()
torch.cuda.empty_cache()
print("Espacio liberado")
```
